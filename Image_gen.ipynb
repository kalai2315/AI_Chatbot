```python
# Install necessary libraries
#!pip install diffusers torch transformers accelerate safetensors --quiet

from diffusers import StableDiffusionPipeline
import torch
from PIL import Image
import IPython.display as display

# Load Stable Diffusion model
model_id = "runwayml/stable-diffusion-v1-5"
pipe = StableDiffusionPipeline.from_pretrained(model_id, torch_dtype=torch.float16)
pipe = pipe.to("cuda" if torch.cuda.is_available() else "cpu")

# Get user input
prompt = input("Enter a topic (e.g., 'solar system diagram'): ")

# Generate the image
print("Generating image, please wait...")
image = pipe(prompt).images[0]

# Display the image
display.display(image)

# Save the image
image.save("generated_image.png")
print("Image saved as 'generated_image.png'.")
```

```python
%%writefile image.py

import streamlit as st
from diffusers import StableDiffusionPipeline
import torch

# Load model with caching
@st.cache_resource
def load_model():
    model_id = "runwayml/stable-diffusion-v1-5"
    pipe = StableDiffusionPipeline.from_pretrained(model_id, torch_dtype=torch.float16)
    return pipe.to("cuda")

pipe = load_model()

# Streamlit UI
st.title(" AI Tutor - Educational Image Generator")
st.write("Enter a topic and get an AI-generated educational image!")

prompt = st.text_input("Enter a topic (e.g., 'solar system diagram')", "")

if st.button("Generate Image"):
    with st.spinner("Generating... "):
        image = pipe(prompt).images[0]
        st.image(image, caption="Generated Image", use_column_width=True)
        st.success(" Image Generated Successfully!")
```
